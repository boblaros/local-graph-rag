## Analysis contract

The experiment is a cumulative cascade: **Native → ER → ER+RR**. The primary contrast is
**ER+RR−Native**, the secondary contrast is **ER−Native**, and the incremental ablation is
**ER+RR−ER**. Uncertainty is estimated by 10,000 paired, question-level bootstrap resamples
stratified by question type (seed 42), matching the evaluated analysis contract.

```python
from __future__ import annotations

import hashlib
import json
import math
import os
import warnings
from pathlib import Path

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)

start = Path.cwd().resolve()
candidates = [start, *start.parents, start / "experiment", start.parent / "experiment"]
PROJECT_ROOT = next(
    path for path in candidates
    if (path / "configs" / "experiment.yaml").is_file() and (path / "runs" / "analysis").is_dir()
)
SECTION_FOLDER = '4.5 Cross-Stage Propagation'
NOTEBOOK_DIR = PROJECT_ROOT / "analysis" / SECTION_FOLDER
OUTPUT_DIR = NOTEBOOK_DIR / "output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
os.environ.setdefault("MPLCONFIGDIR", str(OUTPUT_DIR / ".matplotlib"))

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from cycler import cycler
from IPython.display import Markdown, display
from matplotlib.colors import LinearSegmentedColormap, TwoSlopeNorm
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
from matplotlib.ticker import PercentFormatter

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_colwidth", 100)
pd.set_option("display.float_format", lambda value: f"{value:,.4f}")

OKABE_ITO = {
    "orange": "#E69F00",
    "sky_blue": "#56B4E9",
    "bluish_green": "#009E73",
    "yellow": "#F0E442",
    "blue": "#0072B2",
    "vermillion": "#D55E00",
    "reddish_purple": "#CC79A7",
    "black": "#000000",
}
OKABE_CYCLE = [
    OKABE_ITO["blue"], OKABE_ITO["orange"], OKABE_ITO["bluish_green"],
    OKABE_ITO["vermillion"], OKABE_ITO["sky_blue"],
    OKABE_ITO["reddish_purple"], OKABE_ITO["yellow"], OKABE_ITO["black"],
]
REGIME_ORDER = ["Native", "ER", "ER+RR"]
REGIME_INTERNAL = {
    "Native": "native_lightrag",
    "ER": "advanced_lightrag_er",
    "ER+RR": "advanced_lightrag_er_rr",
}
REGIME_LABEL = {value: key for key, value in REGIME_INTERNAL.items()}
REGIME_COLORS = {
    "Native": OKABE_ITO["black"],
    "ER": OKABE_ITO["blue"],
    "ER+RR": OKABE_ITO["vermillion"],
}
FAMILY_COLORS = {
    "qwen3.5": OKABE_ITO["blue"],
    "qwen3": OKABE_ITO["orange"],
    "gemma3": OKABE_ITO["bluish_green"],
}
QUESTION_COLORS = {
    "comparison": OKABE_ITO["sky_blue"],
    "inference": OKABE_ITO["bluish_green"],
    "temporal": OKABE_ITO["orange"],
    "unanswerable": OKABE_ITO["reddish_purple"],
}
CONTRASTS = [
    ("Primary: ER+RR−Native", "Native", "ER+RR"),
    ("Secondary: ER−Native", "Native", "ER"),
    ("Incremental: ER+RR−ER", "ER", "ER+RR"),
]
CONTRAST_COLORS = {
    "Primary: ER+RR−Native": OKABE_ITO["bluish_green"],
    "Secondary: ER−Native": OKABE_ITO["sky_blue"],
    "Incremental: ER+RR−ER": OKABE_ITO["orange"],
}
CHANGE_CMAP = LinearSegmentedColormap.from_list(
    "okabe_change", [OKABE_ITO["vermillion"], "#FFFFFF", OKABE_ITO["blue"]]
)

plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 320,
    "font.family": "cmr10",
    "mathtext.fontset": "cm",
    "axes.formatter.use_mathtext": True,
    "axes.unicode_minus": False,
    "font.size": 10,
    "axes.labelsize": 10,
    "axes.titlesize": 11,
    "legend.fontsize": 9,
    "axes.grid": True,
    "grid.alpha": 0.20,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.prop_cycle": cycler(color=OKABE_CYCLE),
})

def sha256_file(path: Path) -> str:
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for block in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()

def resolve_registry_path(raw_path: str) -> Path:
    path = Path(raw_path)
    if path.is_file():
        return path
    parts = path.parts
    if "runs" not in parts:
        raise FileNotFoundError(path)
    return PROJECT_ROOT / Path(*parts[parts.index("runs"):])

def read_jsonl(path: Path) -> pd.DataFrame:
    with path.open(encoding="utf-8") as handle:
        records = [json.loads(line) for line in handle if line.strip()]
    return pd.DataFrame(records)

def save_table(frame: pd.DataFrame, stem: str, index: bool = False) -> None:
    csv_path = OUTPUT_DIR / f"{stem}.csv"
    tex_path = OUTPUT_DIR / f"{stem}.tex"
    frame.to_csv(csv_path, index=index)
    latex_frame = frame.replace([np.inf, -np.inf], np.nan)
    tex_path.write_text(
        latex_frame.to_latex(index=index, float_format=lambda value: f"{value:.4f}", na_rep="—"),
        encoding="utf-8",
    )

def save_figure(fig: plt.Figure, stem: str) -> None:
    fig.savefig(OUTPUT_DIR / f"{stem}.png", dpi=320, bbox_inches="tight", facecolor="white")
    fig.savefig(OUTPUT_DIR / f"{stem}.svg", bbox_inches="tight", facecolor="white")

def stratified_mean_ci(
    frame: pd.DataFrame,
    value_col: str,
    *,
    samples: int = 10_000,
    seed: int = 42,
) -> tuple[float, float, float, int]:
    clean_frame = frame[["question_type", value_col]].dropna()
    arrays = [
        group[value_col].to_numpy(dtype=float)
        for _, group in clean_frame.groupby("question_type", sort=True)
        if len(group)
    ]
    if not arrays:
        return (np.nan, np.nan, np.nan, 0)
    observed = np.concatenate(arrays)
    rng = np.random.default_rng(seed)
    bootstrap_sum = np.zeros(samples, dtype=float)
    for values in arrays:
        indices = rng.integers(0, len(values), size=(samples, len(values)))
        bootstrap_sum += values[indices].sum(axis=1)
    bootstrap_means = bootstrap_sum / len(observed)
    lower, upper = np.quantile(bootstrap_means, [0.025, 0.975])
    return float(observed.mean()), float(lower), float(upper), int(len(observed))

def metric_question_level(metric: str, regime: str, builder: str | None = None) -> pd.DataFrame:
    subset = question_metrics.loc[
        question_metrics["graph_regime"].eq(REGIME_INTERNAL[regime]),
        ["builder_key", "question_id", "question_type", metric],
    ].dropna(subset=[metric])
    if builder is not None:
        subset = subset.loc[subset["builder_key"].eq(builder)]
    return (
        subset.groupby(["question_id", "question_type"], as_index=False, observed=True)[metric]
        .mean()
    )

def contrast_question_level(
    metric: str,
    left: str,
    right: str,
    builder: str | None = None,
) -> pd.DataFrame:
    subset = question_metrics[
        ["builder_key", "question_id", "question_type", "graph_regime", metric]
    ].dropna(subset=[metric])
    if builder is not None:
        subset = subset.loc[subset["builder_key"].eq(builder)]
    wide = subset.pivot_table(
        index=["builder_key", "question_id", "question_type"],
        columns="graph_regime",
        values=metric,
        aggfunc="mean",
    )
    left_key, right_key = REGIME_INTERNAL[left], REGIME_INTERNAL[right]
    paired = wide.dropna(subset=[left_key, right_key]).copy()
    paired["difference"] = paired[right_key] - paired[left_key]
    return (
        paired.reset_index()
        .groupby(["question_id", "question_type"], as_index=False, observed=True)["difference"]
        .mean()
    )

def overall_metric_estimate(metric: str, regime: str) -> dict:
    frame = metric_question_level(metric, regime)
    mean, lower, upper, count = stratified_mean_ci(frame, metric)
    return {
        "metric": metric, "regime": regime, "mean": mean,
        "ci_lower": lower, "ci_upper": upper, "question_count": count,
    }

def contrast_metric_estimate(
    metric: str,
    contrast: str,
    left: str,
    right: str,
    builder: str | None = None,
    seed: int = 42,
) -> dict:
    frame = contrast_question_level(metric, left, right, builder)
    mean, lower, upper, count = stratified_mean_ci(
        frame, "difference", seed=seed
    )
    left_frame = metric_question_level(metric, left, builder)
    left_mean = left_frame[metric].mean()
    relative = 100 * mean / left_mean if pd.notna(left_mean) and left_mean != 0 else np.nan
    return {
        "builder_key": builder,
        "contrast": contrast,
        "metric": metric,
        "left_regime": left,
        "right_regime": right,
        "left_mean": left_mean,
        "mean_difference": mean,
        "ci_lower": lower,
        "ci_upper": upper,
        "relative_difference_percent": relative,
        "paired_question_count": count,
    }

print(f"Project root: {PROJECT_ROOT}")
print(f"Output directory: {OUTPUT_DIR}")
```

## Load and validate the evaluated data

The active input registry in the global report is authoritative. This avoids accidentally
including historical variants while still checking every file against its recorded hash,
schema, row count, regime, and paired coverage.

```python
GLOBAL_REPORT_PATH = PROJECT_ROOT / "runs" / "analysis" / "global_experiment_report.34d2c998f8e8143d.json"
global_report = json.loads(GLOBAL_REPORT_PATH.read_text(encoding="utf-8"))
assert global_report["schema_version"] == "2.0.0"
assert global_report["builder_count"] == 12
assert global_report["expected_question_count"] == 120
assert global_report["graph_regimes"] == list(REGIME_LABEL)
assert global_report["bootstrap"] == {
    "cluster_unit": "question_id",
    "confidence_level": 0.95,
    "method": "paired_stratified_percentile_by_question_type",
    "samples": 10000,
    "seed": 42,
    "shared_resample_plan": True,
}

builder_meta = pd.DataFrame([
    {
        "builder_key": row["builder_key"],
        "display_name": row["display_name"],
        "family": row["family"],
        "parameter_billions": row["parameter_billions"],
        "scale_band": row["scale_band"],
    }
    for row in global_report["family_and_scale"]["builder_metrics"]
])
builder_order = builder_meta["builder_key"].tolist()
display_order = builder_meta["display_name"].tolist()
builder_lookup = builder_meta.set_index("builder_key").to_dict("index")

metric_entries = []
for logical_name, metadata in global_report["input_artifacts"].items():
    parts = logical_name.split("/")
    if len(parts) == 2 and parts[1] in REGIME_LABEL:
        metric_entries.append((parts[0], parts[1], metadata))
metric_entries.sort(key=lambda item: (builder_order.index(item[0]), list(REGIME_LABEL).index(item[1])))
assert len(metric_entries) == 36

frames = []
validation_rows = []
required_columns = {
    "question_id", "question_type", "graph_regime", "builder_model",
    "retrieval_hit", "retrieval_recall", "reciprocal_rank",
    "answer_correct", "token_f1", "retrieval_latency_ms",
    "answer_latency_ms", "context_token_count", "answer_total_tokens",
}
for builder_key, regime, metadata in metric_entries:
    path = resolve_registry_path(metadata["path"])
    frame = read_jsonl(path)
    missing = required_columns.difference(frame.columns)
    if missing:
        raise ValueError(f"{path} is missing required fields: {sorted(missing)}")
    actual_hash = sha256_file(path)
    valid = (
        actual_hash == metadata["sha256"]
        and len(frame) == metadata["record_count"] == 120
        and set(frame["schema_version"]) == {metadata["schema_version"]} == {"3.0.0"}
        and set(frame["graph_regime"]) == {regime}
    )
    if not valid:
        raise ValueError(f"Artifact validation failed: {path}")
    frame["builder_key"] = builder_key
    frames.append(frame)
    validation_rows.append({
        "builder_key": builder_key,
        "regime": REGIME_LABEL[regime],
        "relative_path": path.relative_to(PROJECT_ROOT).as_posix(),
        "records": len(frame),
        "schema_version": metadata["schema_version"],
        "sha256_verified": actual_hash == metadata["sha256"],
    })

question_metrics = pd.concat(frames, ignore_index=True)
question_metrics = question_metrics.merge(builder_meta, on="builder_key", how="left", validate="many_to_one")
question_metrics["regime"] = question_metrics["graph_regime"].map(REGIME_LABEL)
question_metrics["regime"] = pd.Categorical(
    question_metrics["regime"], categories=REGIME_ORDER, ordered=True
)
question_metrics["builder_key"] = pd.Categorical(
    question_metrics["builder_key"], categories=builder_order, ordered=True
)
source_validation = pd.DataFrame(validation_rows)

coverage = (
    question_metrics.groupby(["builder_key", "regime"], observed=True)["question_id"]
    .nunique().unstack()
)
if question_metrics.shape[0] != 4_320 or not coverage.eq(120).all().all():
    raise ValueError(f"Unexpected evaluated coverage: rows={len(question_metrics)}")
if question_metrics[["builder_key", "regime", "question_id"]].duplicated().any():
    raise ValueError("Duplicate builder–regime–question rows detected")

save_table(source_validation, "source_validation")
display(source_validation)
print(f"Validated {len(source_validation)} source files and {len(question_metrics):,} evaluated rows.")
print(f"Question types: {question_metrics['question_type'].value_counts().to_dict()}")
```

## Construct the builder-level cross-stage dataset

Graph construction and post-processing metrics come from the report’s artifact-level records. Retrieval and
QA effects use the same builder-matched cascade contrasts as Sections 4.3–4.4. The unit is the builder (n=12);
associations are descriptive and cannot establish causal mediation.

```python
artifact_rows = {
    row["builder_key"]: row
    for row in global_report["family_and_scale"]["artifact_level"]["builder_metrics"]
}
performance_rows = {
    row["builder_key"]: row
    for row in global_report["family_and_scale"]["builder_metrics"]
}
cross_rows = []
for builder in builder_order:
    meta = builder_lookup[builder]
    artifact = artifact_rows[builder]["metrics"]
    performance = performance_rows[builder]["metrics"]
    cross_rows.append({
        "builder_key": builder,
        **meta,
        "native_nodes": artifact["native_topology.nodes"],
        "native_edges": artifact["native_topology.edges"],
        "native_largest_component_share": artifact["native_topology.largest_component_share"],
        "primary_node_change": artifact["primary_topology_effect.nodes"],
        "primary_edge_change": artifact["primary_topology_effect.edges"],
        "primary_average_degree_change": artifact["primary_topology_effect.average_degree"],
        "primary_lcc_share_change": artifact["primary_topology_effect.largest_component_share"],
        "secondary_node_change": artifact["secondary_topology_effect.nodes"],
        "secondary_edge_change": artifact["secondary_topology_effect.edges"],
        "rr_edge_change": artifact["incremental_ablation_topology_effect.edges"],
        "rr_lcc_share_change": artifact["incremental_ablation_topology_effect.largest_component_share"],
        "extraction_total_tokens": artifact["extraction.efficiency.total_tokens"],
        "extracted_mentions": artifact["extraction.mentions"],
        "extracted_relations": artifact["extraction.relations"],
        "er_merge_count": artifact["entity_resolution.merge_count"],
        "er_candidate_count": artifact["entity_resolution.candidate_count"],
        "er_runtime_seconds": artifact["entity_resolution.er_runtime_seconds"],
        "er_judge_tokens": artifact["entity_resolution.judge_tokens"],
        "primary_retrieval_hit_effect": performance["retrieval_hit"]["primary_effect"],
        "primary_retrieval_recall_effect": performance["retrieval_recall"]["primary_effect"],
        "primary_mrr_effect": performance["reciprocal_rank"]["primary_effect"],
        "primary_answer_correct_effect": performance["answer_correct"]["primary_effect"],
        "primary_token_f1_effect": performance["token_f1"]["primary_effect"],
        "secondary_retrieval_hit_effect": performance["retrieval_hit"]["secondary_effect"],
        "secondary_answer_correct_effect": performance["answer_correct"]["secondary_effect"],
        "incremental_retrieval_hit_effect": performance["retrieval_hit"]["incremental_ablation_effect"],
        "incremental_answer_correct_effect": performance["answer_correct"]["incremental_ablation_effect"],
        "native_retrieval_hit": performance["retrieval_hit"]["native"],
        "native_answer_correct": performance["answer_correct"]["native"],
    })
cross_stage = pd.DataFrame(cross_rows)
cross_stage["er_merge_rate_per_mention"] = cross_stage["er_merge_count"] / cross_stage["extracted_mentions"]
cross_stage["rr_edge_change_percent_of_er_edges"] = 100 * cross_stage["rr_edge_change"] / (
    cross_stage["native_edges"] + cross_stage["secondary_edge_change"]
)
if len(cross_stage) != 12 or cross_stage.isna().any().any():
    raise ValueError("Cross-stage dataset is incomplete")
save_table(cross_stage, "cross_stage_builder_dataset")
display(cross_stage)
```

## Planned descriptive associations with bootstrap uncertainty

Spearman rank correlations are robust to non-linear monotonic patterns and scale differences. Intervals use
10,000 builder-level bootstrap resamples. With n=12, wide intervals and unstable signs are scientifically
informative; no p-values or causal claims are attached.

```python
def spearman_rho(x, y):
    pair = pd.DataFrame({"x": x, "y": y}).dropna()
    if len(pair) < 3 or pair["x"].nunique() < 2 or pair["y"].nunique() < 2:
        return np.nan
    return pair["x"].rank().corr(pair["y"].rank())

def bootstrap_spearman(x, y, samples=10_000, seed=42):
    pair = pd.DataFrame({"x": x, "y": y}).dropna().reset_index(drop=True)
    observed = spearman_rho(pair["x"], pair["y"])
    rng = np.random.default_rng(seed)
    boot = np.empty(samples)
    for index in range(samples):
        sample = pair.iloc[rng.integers(0, len(pair), len(pair))]
        boot[index] = spearman_rho(sample["x"], sample["y"])
    valid = boot[np.isfinite(boot)]
    if not len(valid):
        return observed, np.nan, np.nan
    return observed, *np.quantile(valid, [0.025, 0.975])

association_plan = [
    ("Primary edge change", "primary_edge_change", "Primary retrieval-hit effect", "primary_retrieval_hit_effect"),
    ("Primary average-degree change", "primary_average_degree_change", "Primary retrieval-recall effect", "primary_retrieval_recall_effect"),
    ("Primary LCC-share change", "primary_lcc_share_change", "Primary retrieval-hit effect", "primary_retrieval_hit_effect"),
    ("Primary retrieval-hit effect", "primary_retrieval_hit_effect", "Primary answer-correctness effect", "primary_answer_correct_effect"),
    ("Primary retrieval-recall effect", "primary_retrieval_recall_effect", "Primary Token-F1 effect", "primary_token_f1_effect"),
    ("ER node change", "secondary_node_change", "ER retrieval-hit effect", "secondary_retrieval_hit_effect"),
    ("ER merge rate", "er_merge_rate_per_mention", "ER retrieval-hit effect", "secondary_retrieval_hit_effect"),
    ("RR edge change", "rr_edge_change", "RR retrieval-hit effect", "incremental_retrieval_hit_effect"),
    ("RR LCC-share change", "rr_lcc_share_change", "RR answer-correctness effect", "incremental_answer_correct_effect"),
    ("Native graph nodes", "native_nodes", "Native retrieval hit", "native_retrieval_hit"),
    ("Parameter count", "parameter_billions", "Primary retrieval-hit effect", "primary_retrieval_hit_effect"),
    ("Parameter count", "parameter_billions", "Primary answer-correctness effect", "primary_answer_correct_effect"),
]
association_rows = []
for index, (x_label, x_col, y_label, y_col) in enumerate(association_plan):
    rho, lower, upper = bootstrap_spearman(
        cross_stage[x_col], cross_stage[y_col], seed=42 + index
    )
    association_rows.append({
        "predictor": x_label, "predictor_field": x_col,
        "outcome": y_label, "outcome_field": y_col,
        "spearman_rho": rho, "ci_lower": lower, "ci_upper": upper,
        "builder_count": len(cross_stage),
        "absolute_rho": abs(rho),
    })
associations = pd.DataFrame(association_rows).sort_values("absolute_rho", ascending=False)
save_table(associations, "planned_cross_stage_associations")
display(associations)
```

## Figure 4.5.1 — Cross-stage association matrix

Rows are graph or process descriptors; columns are downstream treatment effects. The matrix is an exploratory
map of monotonic relationships, not evidence that graph changes caused retrieval or answer changes.

```python
predictors = {
    r"Primary node $\Delta$": "primary_node_change",
    r"Primary edge $\Delta$": "primary_edge_change",
    r"Primary degree $\Delta$": "primary_average_degree_change",
    r"Primary LCC share $\Delta$": "primary_lcc_share_change",
    "ER merge rate": "er_merge_rate_per_mention",
    r"RR edge $\Delta$": "rr_edge_change",
}
outcomes = {
    r"Retrieval hit $\Delta$": "primary_retrieval_hit_effect",
    r"Recall $\Delta$": "primary_retrieval_recall_effect",
    r"MRR $\Delta$": "primary_mrr_effect",
    r"Correctness $\Delta$": "primary_answer_correct_effect",
    r"Token F1 $\Delta$": "primary_token_f1_effect",
}
matrix = pd.DataFrame(
    {
        outcome_label: [
            spearman_rho(cross_stage[predictor_col], cross_stage[outcome_col])
            for predictor_col in predictors.values()
        ]
        for outcome_label, outcome_col in outcomes.items()
    },
    index=predictors,
)
save_table(matrix.reset_index(names="predictor"), "cross_stage_spearman_matrix")

fig, ax = plt.subplots(figsize=(9.8, 6.2), constrained_layout=True)
image = ax.imshow(matrix, aspect="auto", cmap=CHANGE_CMAP, vmin=-1, vmax=1)
ax.set_xticks(range(matrix.shape[1]), matrix.columns, rotation=25, ha="right")
ax.set_yticks(range(matrix.shape[0]), matrix.index)
for row in range(matrix.shape[0]):
    for column in range(matrix.shape[1]):
        ax.text(column, row, f"{matrix.iloc[row, column]:+.2f}", ha="center", va="center")
colorbar = fig.colorbar(image, ax=ax, pad=0.02)
colorbar.set_label("Spearman rank correlation")
save_figure(fig, "01_cross_stage_association_matrix")
plt.show()
```

## Figure 4.5.2 — Retrieval-to-generation propagation

Each point is one builder. The quadrants distinguish joint gains, retrieval-only gains, QA-only gains, and joint
deterioration. This directly tests whether better access to evidence consistently propagates to correct answers.
Labels identify the eight most extreme joint positions to keep the figure readable.

```python
fig, ax = plt.subplots(figsize=(9.2, 7), constrained_layout=True)
propagation_plot = cross_stage.copy()
for field in ["primary_retrieval_hit_effect", "primary_answer_correct_effect"]:
    spread = propagation_plot[field].std(ddof=0)
    propagation_plot[f"{field}_z"] = (
        (propagation_plot[field] - propagation_plot[field].median()) / spread
        if spread else 0
    )
propagation_plot["label_score"] = (
    propagation_plot["primary_retrieval_hit_effect_z"].abs()
    + propagation_plot["primary_answer_correct_effect_z"].abs()
)
labelled = propagation_plot.nlargest(8, "label_score")
for family, group in propagation_plot.groupby("family"):
    ax.scatter(
        group["primary_retrieval_hit_effect"] * 100,
        group["primary_answer_correct_effect"] * 100,
        s=80, color=FAMILY_COLORS[family], label=family,
        edgecolor="white", linewidth=0.8,
    )
for position, row in enumerate(labelled.sort_values("primary_answer_correct_effect").itertuples()):
    offset = (5, 5 if position % 2 == 0 else -11)
    ax.annotate(
        row.display_name,
        (row.primary_retrieval_hit_effect * 100, row.primary_answer_correct_effect * 100),
        xytext=offset, textcoords="offset points", fontsize=7.5,
    )
ax.axhline(0, color=OKABE_ITO["black"], linewidth=1)
ax.axvline(0, color=OKABE_ITO["black"], linewidth=1)
ax.set_xlabel("Primary retrieval-hit effect (percentage points)")
ax.set_ylabel("Primary answer-correctness effect (percentage points)")
ax.legend(title="Model family", frameon=False)
rho = spearman_rho(cross_stage["primary_retrieval_hit_effect"], cross_stage["primary_answer_correct_effect"])
ax.text(0.02, 0.97, rf"Spearman $\rho$ = {rho:+.2f}", transform=ax.transAxes, va="top")
save_figure(fig, "02_retrieval_to_generation_propagation")
plt.show()
```

## Figure 4.5.5 — Graph-to-retrieval case trajectories

The three panels follow two structurally atypical builders through the cumulative
**Native → ER → ER+RR** cascade, connecting graph fragmentation to complete-chain recall
and retrieval hit rate. No aggregate inference is made from these two cases; they illustrate
how post-processing can improve multi-hop retrieval through different stages.


```python
CASE_BUILDERS = {
    "qwen3_8b": {
        "label": "Qwen3 8B",
        "color": OKABE_ITO["orange"],
        "marker": "o",
    },
    "gemma3_270m": {
        "label": "Gemma 3 270M",
        "color": OKABE_ITO["bluish_green"],
        "marker": "s",
    },
}

topology_prefix = {
    "Native": "native_topology",
    "ER": "er_topology",
    "ER+RR": "er_rr_topology",
}

case_rows = []
for builder_key, builder_style in CASE_BUILDERS.items():
    artifact = artifact_rows[builder_key]["metrics"]
    for regime in REGIME_ORDER:
        subset = question_metrics.loc[
            question_metrics["builder_key"].astype(str).eq(builder_key)
            & question_metrics["regime"].astype(str).eq(regime)
        ]
        answerable = subset.loc[subset["answerable"]]
        prefix = topology_prefix[regime]
        case_rows.append({
            "builder_key": builder_key,
            "display_name": builder_style["label"],
            "regime": regime,
            "isolated_nodes_percent": 100 * artifact[f"{prefix}.isolates"] / artifact[f"{prefix}.nodes"],
            "retrieval_hit_percent": 100 * answerable["retrieval_hit"].mean(),
            "complete_chain_recall_at_5_percent": 100 * answerable["complete_chain_recall_at_5"].mean(),
            "complete_chain_count": int(answerable["complete_chain_recall_at_5"].sum()),
            "answerable_count": int(len(answerable)),
        })

case_trajectories = pd.DataFrame(case_rows)
case_trajectories["regime"] = pd.Categorical(
    case_trajectories["regime"], categories=REGIME_ORDER, ordered=True
)
case_trajectories = case_trajectories.sort_values(["builder_key", "regime"])
assert len(case_trajectories) == 6
assert case_trajectories["answerable_count"].eq(90).all()
assert not case_trajectories.isna().any().any()
save_table(case_trajectories, "05_cross_stage_case_trajectory_values")

x = np.arange(len(REGIME_ORDER))
fig, axes = plt.subplots(1, 3, figsize=(13.2, 4.5), constrained_layout=True)

def plot_builder_trajectories(ax, metric):
    for builder_key, builder_style in CASE_BUILDERS.items():
        values = (
            case_trajectories.loc[case_trajectories["builder_key"].eq(builder_key)]
            .set_index("regime")
            .reindex(REGIME_ORDER)[metric]
            .to_numpy(dtype=float)
        )
        ax.plot(
            x, values,
            color=builder_style["color"],
            marker=builder_style["marker"],
            markersize=7.5,
            linewidth=2.2,
            label=builder_style["label"],
        )

ax = axes[0]
plot_builder_trajectories(ax, "isolated_nodes_percent")
ax.set_ylabel("Isolated nodes (%)")
ax.set_ylim(35, 90)
ax.set_yticks([35, 46, 57, 68, 79, 90])
ax.legend(loc="upper right", frameon=False)

ax = axes[1]
plot_builder_trajectories(ax, "complete_chain_recall_at_5_percent")
ax.set_ylabel("Complete-chain recall@5 (%)")
ax.set_ylim(0, 8)
for builder_key, builder_style in CASE_BUILDERS.items():
    rows = (
        case_trajectories.loc[case_trajectories["builder_key"].eq(builder_key)]
        .set_index("regime")
        .reindex(REGIME_ORDER)
    )
    for idx, row in enumerate(rows.itertuples()):
        offset = 0.22 if builder_key == "qwen3_8b" else -0.42
        ax.text(
            idx,
            row.complete_chain_recall_at_5_percent + offset,
            f"{row.complete_chain_count}/{row.answerable_count}",
            color=builder_style["color"],
            fontsize=8,
            ha="center",
        )

ax = axes[2]
plot_builder_trajectories(ax, "retrieval_hit_percent")
ax.set_ylabel("Retrieval hit rate (%)")
ax.set_ylim(34, 51)
ax.set_yticks([34, 37, 41, 44, 48, 51])

for ax in axes:
    ax.set_xticks(x, REGIME_ORDER)
    ax.set_xlim(-0.05, len(REGIME_ORDER) - 0.95)
    ax.grid(axis="x", visible=False)

save_figure(fig, "05_cross_stage_case_trajectories")
display(case_trajectories)
plt.show()
```

## Figure 4.5.6 — Retrieval–generation funnel

The funnel aggregates all 12 builders within each regime. Because every builder contributes
the same 90 answerable questions, the pooled 1,080 observations also weight builders equally.
The final stage counts correct answers only within official gold-document-hit cases, preserving
a nested retrieval-to-generation pathway. Correct answers without an official hit are reported
in the exported table but excluded from the funnel.


```python
funnel_rows = []
for regime in REGIME_ORDER:
    answerable = question_metrics.loc[
        question_metrics["regime"].astype(str).eq(regime)
        & question_metrics["answerable"].astype(bool)
    ].copy()
    hit = answerable["retrieval_hit"].astype(bool)
    correct = answerable["answer_correct"].astype(bool)
    total_count = int(len(answerable))
    hit_count = int(hit.sum())
    correct_with_hit_count = int((hit & correct).sum())
    correct_without_hit_count = int((~hit & correct).sum())
    funnel_rows.append({
        "regime": regime,
        "answerable_evaluations": total_count,
        "gold_document_hit": hit_count,
        "correct_with_gold_hit": correct_with_hit_count,
        "correct_without_gold_hit": correct_without_hit_count,
        "gold_hit_percent": 100 * hit_count / total_count,
        "correct_with_hit_percent_of_total": 100 * correct_with_hit_count / total_count,
        "hit_to_correct_conversion_percent": 100 * correct_with_hit_count / hit_count,
    })

retrieval_generation_funnel = pd.DataFrame(funnel_rows)
assert retrieval_generation_funnel["answerable_evaluations"].eq(1080).all()
assert (retrieval_generation_funnel["correct_with_gold_hit"] <= retrieval_generation_funnel["gold_document_hit"]).all()
save_table(retrieval_generation_funnel, "06_retrieval_generation_funnel_values")

fig, axes = plt.subplots(1, 3, figsize=(13.2, 4.8), constrained_layout=True)
stage_colors = [OKABE_ITO["sky_blue"], OKABE_ITO["blue"], OKABE_ITO["vermillion"]]
for ax, row in zip(axes, retrieval_generation_funnel.itertuples(index=False)):
    counts = np.array([
        row.answerable_evaluations, row.gold_document_hit, row.correct_with_gold_hit
    ], dtype=float)
    widths = 0.48 * counts / counts[0]
    levels = np.array([2.0, 1.0, 0.0])
    for idx in range(2):
        ax.fill(
            [-widths[idx], widths[idx], widths[idx + 1], -widths[idx + 1]],
            [levels[idx], levels[idx], levels[idx + 1], levels[idx + 1]],
            color=stage_colors[idx], alpha=0.88, edgecolor="white", linewidth=2.2,
        )
    ax.plot([-widths[2], widths[2]], [0, 0], color=stage_colors[2], linewidth=8, solid_capstyle="butt")
    labels = [
        f"Answerable evaluations\n{int(counts[0]):,} (100%)",
        f"Gold-document hit\n{int(counts[1]):,} ({100 * counts[1] / counts[0]:.1f}%)",
        f"Correct within hit cases\n{int(counts[2]):,} ({100 * counts[2] / counts[0]:.1f}%)",
    ]
    for y, label in zip(levels, labels):
        ax.text(0, y + (0.12 if y else -0.17), label, ha="center", va="center", fontsize=9)
    ax.text(
        0, -0.55, f"Gold-hit to correct conversion: {row.hit_to_correct_conversion_percent:.1f}%",
        ha="center", va="center", fontsize=9.5, fontweight="bold",
    )
    ax.set_title(row.regime, fontweight="bold")
    ax.set_xlim(-0.56, 0.56)
    ax.set_ylim(-0.78, 2.32)
    ax.axis("off")

save_figure(fig, "06_retrieval_generation_funnel")
display(retrieval_generation_funnel)
plt.show()
```
